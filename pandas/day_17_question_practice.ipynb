###`Q-1:`
i. Create a DataFrame birds from the above dictionary data which has the index labels.

ii. Display basic information about the dataFrame.

````python
# code here
data= {'birds': ['Cranes', 'Cranes', 'plovers', 'spoonbills', 'spoonbills', 'Cranes', 'plovers', 'Cranes', 'spoonbills', 'spoonbills', 'Cranes'],
        'age': [3.5, 4, 1.5, np.nan, 6, 3, 5.5, np.nan, 8, 4, 3.5], 'visits': [2, 4, 3, 4, 3, 4, 2, 2, 3, 2, 2],
        'priority': ['yes', 'yes', 'no', np.nan, 'no', 'no', 'no', 'yes', 'no', 'no','yes']}
labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k']
birds=pd.DataFrame(data,index=labels)
birds

````

###`Q-3:`
i. Select all rows with nan values in age and visits column.

ii. Fill nan with respective series mode value.

````python
# code here
birds[birds['age'].isna() | birds['visits'].isna()]
````

````python
age_mode = birds['age'].mode().iloc[0]
visits_mode = birds['visits'].mode().iloc[0]
````

````python
birds['age'] = birds['age'].fillna(age_mode)
birds['visits'] = birds['visits'].fillna(visits_mode)
````

###`Q-6` Write a code which can display the bar chart of top 5 teams who have played maximum number of matches in the IPL.
>  Hint: Be careful the data is divided in 2 different cols(Team 1 and Team 2)


````python
# code here
match_played=ipl['Team1'].value_counts()+ipl['Team2'].value_counts()
````

````python
match_played.sort_values(ascending=False).head().plot(kind='bar')
````
