
###`Q-5:` Find batsman in below category-
* Highest score while chasing
* Best Strike rate while chasing and have faced 50+ balls


> Chasing mean team batting in second inning

````python
# code here
matches=pd.read_csv('ipl_deliveries.csv')
matches.head()
````

````python
matches[matches['innings']==2].groupby(['ID','batter'])['batsman_run'].sum().sort_values(ascending=False).head()
````

````python
result = (matches[matches['innings'] == 2].groupby(['batter','ID']).agg(runs=('batsman_run', 'sum'),balls=('batsman_run', 'count') ).reset_index())
result['strike_rate'] = result['runs'] / result['balls'] * 100
result[result['balls'] > 50].sort_values('strike_rate', ascending=False).head(1)
````

###`Q-8:` Make a dataframe for all batting pairs played together.
```
Batsman1 Batsman2 Runs Avg StrikeRate
```

> Just to ease this question you can count wide-balls for strike rate.

````python
# code here
match=matches[['batter','non-striker','batting','total_run']]
match
````

````python
result=match.groupby('batting').agg(runs=('total_run','sum'),total_ball=('total_run','count')).reset_index()
result.head()
````

````python
result['batting_strike']=((result['runs']/result['total_ball'])*100).round(2)
result
````
