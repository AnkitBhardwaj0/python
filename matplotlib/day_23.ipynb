### **`Problem-1:`** Draw a line plot of which, the x-axis is the "Year" and the y-axis is sum of "PM2.5" of two countries Iran and China.

```python
# code here
df = pd.read_csv('https://tinyurl.com/2fe6vz4u')
df
```

```python
iran=df[df['Country']=='Iran'].groupby('Year')['PM2.5'].sum()

china=df[df['Country']=='China'].groupby('Year')['PM2.5'].sum()
```

```python
iran
```

```python
china
```

```python
iran.plot(x='Year',y='PM2.5',kind='line')
china.plot(x='Year',y='PM2.5',kind='line')
plt.xlabel('year')
plt.ylabel('PM2.5')
plt.title('PM2.5 over year')
plt.show()
```

```python
iran.plot(x='Year',y='PM2.5',kind='line')
china.plot(x='Year',y='PM2.5',kind='line')
plt.xlabel('year')
plt.ylabel('PM2.5')
plt.title('PM2.5 over year')
plt.grid()
plt.show()
```

```python
iran.plot(x='Year',y='PM2.5',kind='line')
china.plot(x='Year',y='PM2.5',kind='line')
plt.xlabel('year')
plt.ylabel('PM2.5')
plt.title('PM2.5 over year')
plt.legend()
plt.grid()
plt.show()
```

```python
iran.plot(x='Year',y='PM2.5',kind='line',color='orange',label='iran',linestyle='dotted')
china.plot(x='Year',y='PM2.5',kind='line',color='red',label='china',linestyle='dashed')
plt.xlabel('year')
plt.ylabel('PM2.5')
plt.title('PM2.5 over year')
plt.grid()
plt.show()
```

```python
iran.plot(x='Year',y='PM2.5',kind='line',color='orange',label='iran',linestyle='dotted')
china.plot(x='Year',y='PM2.5',kind='line',color='red',label='china',linestyle='dashed')
plt.xlabel('year')
plt.ylabel('PM2.5')
plt.title('PM2.5 over year')
plt.xticks(df['Year'])
plt.grid()
plt.show()
```

```python
iran.plot(x='Year',y='PM2.5',kind='line',label='iran',color='orange',linestyle='dotted')
china.plot(x='Year',y='PM2.5',kind='line',label='china',color='red',linestyle='dashed')
plt.xlabel('year')
plt.ylabel('PM2.5')
plt.title('PM2.5 over year')
plt.xticks(df['Year'].value_counts().index)
plt.legend()
plt.grid()
plt.show()
```
