```python
class Atm:
    def __init__(self):
        self.balance=0
        self.pin=None
        self.create_pin()
    def create_pin(self):
        while True:
            if self.pin is None:
                self.pin=input("set your pin")
                if len(self.pin)==6 and self.pin.isdigit():
                    break
                else:
                    print("pin must be 6 digits")
        self.menu()
    def menu(self):
        print("""welcome to the atm
                         1.press 1 to deposit
                         2. press 2 to change pin
                         3. press 3 to check balance
                         4. press 4 to withdraw
                         5. anything else to exit
                         """)
       
        while(True):
            user_input=input(""" welcome to the atm
                             enter your choice
                            """)
            if user_input not in ["1","2","3","4"]:
                print("thank you for using the atm")
                break

            user_pin=input("enter your pin")
            if user_pin!=self.pin:
                print("invalid pin")
                continue
            if user_input=="1":
                self.deposit()
            elif user_input=="2":
                self.change_pin()
            elif user_input=="3":
                self.check_balance()
            elif user_input=="4":
                self.withdraw()
            
    def deposit(self):
        self.balance+=int(input("enter amount to deposit"))
        
    def change_pin(self):
        old_pin=input("enter old pin")
        if old_pin==self.pin:
            new_pin=input("enter new pin ")
            re_enter_new_pin=input("re enter new pin")
            if new_pin==re_enter_new_pin and len(new_pin)==6 and new_pin.isdigit():
                self.pin=new_pin
            else:
                print("pin could not be changed")
        else:
            print("invalid old pin")
        
    def check_balance(self):
        print(f"your balance is {self.balance} \n")
        
    def withdraw(self):
        amount=int(input("enter amount to withdraw"))
        if amount<=self.balance:
            print("please collect your cash")
            self.balance-=amount 
        else:
            print("insufficient balance")
```
