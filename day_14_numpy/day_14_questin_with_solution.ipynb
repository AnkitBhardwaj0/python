###`Q-1:` Find the nearest element in the array to a given integer.
```
Ex:-
a=23 and array - [10 17 24 31 38 45 52 59].
Nearest element is 24
```
Hint: Read about this function `argmin()`

````python
import numpy as np
````

````python
# code here
a=np.array([10,17,24,31,38 ,45, 52 ,59])
difference = np.abs(23 - a)
index=np.argmin(difference)
nearest_element=a[index]
print(f"nearest element {nearest_element}")
````

###`Q-4:` Your are given an array which is havig some nan value. You job is to fill those nan values with most common element in the array.
```
arr=np.array([[1,2,np.nan],[4,2,6],[np.nan,np.nan,5]])

```

````python
# code here
arr=np.array([[1,2,np.nan],[4,2,6],[np.nan,np.nan,5]])
print(arr,"\n")
b=arr[~np.isnan(arr)]
print(b,"\n")
value,count=np.unique(b,return_counts=True)
index=np.argmax(count)

a=np.where(np.isnan(arr),value[index],arr)
print(a)
````

### `Q-9:` Write a program implement Boradcasting Rule to check if two array can be added or not.
Given tuples of shapes.
```
shape of a- (3,2,2)
shape of b- (2,2)

check_broadcast(a, b) -> return Boolean (True if can broadcasted, False other wise.)
```

````python
# code here
def check_broadcast(shape_a, shape_b):
    for a, b in zip(shape_a[::-1], shape_b[::-1]):    
        if a != b and a != 1:
            return False
    return True 
shape_a = (3, 2, 2)
shape_b = (2, 2)
print(check_broadcast(shape_a, shape_b))
````

````python
def check_broadcast(shape_a, shape_b):
    i = len(shape_a) - 1
    j = len(shape_b) - 1
    while i >= 0 and j >= 0:

        if (
            shape_a[i] != shape_b[j]
            and shape_a[i] != 1
            and shape_b[j] != 1
        ):
            return False
        i -= 1
        j -= 1
    return True
shape_a = (3, 2)
shape_b = (2, 2)
print(check_broadcast(shape_a, shape_b))
````
